# Sincronizador de base de datos
Descarga los archivos json de la tabla correspondiente, para luego solicitar la ejecución de la solicitud de sincronización.

2026-01-17

```python
import os
import requests
```

```python
table_prefix = 'flipbook_contenido-f432d135d8936657e5f89013-64'

# separar secciones en table y prefix separados por "_"
table, prefix, qty_files = table_prefix.split('-')

print(table)
print(prefix)
print(qty_files)

url_remote = 'https://www.plataformaenlinea.com/2017/'
local_url = "http://localhost/editores/"
path_download = f'D:\\xampp\\htdocs\\editores\\content\\database\\{table}\\'
```

```python
# Generar la lista de nombres de archivos para descargar
def get_files_list(prefix, qty_files):
    qty_files = int(qty_files)
    return [f"{prefix}_{i}.json" for i in range(1, qty_files + 1)]
```

```python
# Generar los archivos y obtener la lista para descargar
files_list = get_files_list(prefix, qty_files)

# Recorrer files_list para descargar cada archivo
for file_name in files_list:
    url_download = f"{url_remote.rstrip('/')}/content/database/{table}/{file_name}"
    file_path = os.path.join(path_download, file_name)

    try:
        response = requests.get(url_download, timeout=15)
        response.raise_for_status()  # Verifica que la solicitud fue exitosa

        os.makedirs(path_download, exist_ok=True)

        with open(file_path, 'wb') as file:
            file.write(response.content)

        print(f"✅ Archivo descargado y guardado en: {file_path}")

    except requests.exceptions.RequestException as e:
        print(f"❌ Error al descargar el archivo: {e}")
```
